```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("saprk_job_salses_csv").getOrCreate()
```

```python
df = spark.read.csv('/content/sample_data/sales.csv', header=True, inferSchema=True)
```

```python
print("\n print schema ")
df.printSchema()
```

```python
print("\n sample rows from sales dataFrame")
print("\n equivalent of sql: select * from table limit 5")
df.show(5)
```

```python
print("\n  select list of columns")
print("\n equivalent of sql: select order_id, customer_id from table")
df.select("order_id", "customer_id").show(5)

print("\n using method-2")
df.select(df["order_id"], df["customer_id"]).show(5)

print("\n using method-3")
from pyspark.sql.functions import col
df.select(col("order_id"), col("customer_id")).show(5)

```

```python
print("\n using method-4")
from pyspark.sql.functions import col
df.select(col("order_id").alias("o_id"), col("customer_id").alias("cust_id")).show(5)
```

```python
print("\n method-5")
df.selectExpr("order_id as o_id", "customer_id as cust_id").show(5)
```

```python
print("\n method-6 withcolumn")
from pyspark.sql.functions import col
df.withColumn("order_id_rn", col("order_id").cast("string")).printSchema()
```

```python
df.filter(df.Payment_Mode == "UPI").show(10)
```

```python
df.filter( (df.Payment_Mode == "UPI") & (df.Order_Status == "Delivered")).show(10)
```

```python
from pyspark.sql.functions import sum
from pyspark.sql.types import DecimalType

df.groupBy("Payment_Mode").agg(sum("Total_Amount").cast(DecimalType(18, 2)).alias("Total_Sales")).show()
```

```python
print(spark.sparkContext.uiWebUrl)
```

```python
from pyspark.sql.functions import avg

df.groupBy("Payment_Mode").agg(avg("Total_Amount").alias("Average_Order_Value")).show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate the average order value per payment mode
avg_order_value_df = df.groupBy("Payment_Mode").agg(avg("Total_Amount").alias("Average_Order_Value"))

# Convert to Pandas DataFrame for plotting
avg_order_value_pd = avg_order_value_df.toPandas()

# Create the bar chart
plt.figure(figsize=(10, 6))
sns.barplot(x='Payment_Mode', y='Average_Order_Value', data=avg_order_value_pd, palette='viridis')
plt.title('Average Order Value per Payment Mode')
plt.xlabel('Payment Mode')
plt.ylabel('Average Order Value')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
