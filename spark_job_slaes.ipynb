```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("saprk_job_salses_csv").getOrCreate()
```

```python
df = spark.read.csv('/content/sample_data/sales.csv', header=True, inferSchema=True)
```

```python
print("\n print schema ")
df.printSchema()
```

```python
print("\n sample rows from sales dataFrame")
print("\n equivalent of sql: select * from table limit 5")
df.show(5)
```

```python
print("\n  select list of columns")
print("\n equivalent of sql: select order_id, customer_id from table")
df.select("order_id", "customer_id").show(5)

print("\n using method-2")
df.select(df["order_id"], df["customer_id"]).show(5)

print("\n using method-3")
from pyspark.sql.functions import col
df.select(col("order_id"), col("customer_id")).show(5)

```

```python
print("\n using method-4")
from pyspark.sql.functions import col
df.select(col("order_id").alias("o_id"), col("customer_id").alias("cust_id")).show(5)
```

```python
print("\n method-5")
df.selectExpr("order_id as o_id", "customer_id as cust_id").show(5)
```

```python
print("\n method-6 withcolumn")
from pyspark.sql.functions import col
df.withColumn("order_id_rn", col("order_id").cast("string")).printSchema()
```

```python
df.filter(df.Payment_Mode == "UPI").show(10)
```

```python
df.filter( (df.Payment_Mode == "UPI") & (df.Order_Status == "Delivered")).show(10)
```
